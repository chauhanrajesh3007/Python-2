```python
import requests
from bs4 import BeautifulSoup
url="https://www.politifact.com/"
response=requests.get(url)#step-1
#print(response.status_code)
response=response.text#step-2
soup=BeautifulSoup(response,"html.parser")#step-3
#print(soup)
#print(soup.prettify)

source=[]
statement=[]
link=[]
date=[]
d1=[]
a=soup.find("div",class_="embla")
b=soup.find_all("div",class_="embla__slide caro-listing__item")
for i in b:
    post=i.find("div",class_="m-statement__meta").find("a").text.strip()
    source.append(post)
    s=i.find("div",class_="m-statement__quote").find("a").text.strip()
    statement.append(s)
    l=i.find("div",class_="m-statement__quote").find("a")["href"]
    link.append("https://www.politifact.com"+l)
    d=i.find("div",class_="m-statement__desc").text.strip()
    l_d=d.split()
    l1=(" ".join(l_d[2:5]))
    date.append(l1)

import pandas as pd
df=pd.DataFrame({"source":source,"statement":statement,"link":link,"Date":date,})
df
```

# create dataframe with columns car name,rating,review count,price,dealer name,mileage

```python
import numpy as np
name=[]
mileage=[]
dealer_name=[]
price=[]
review=[]
rating=[]
a=soup.find("div",class_="sds-page-section__content")
b=soup.find_all("div",class_="vehicle-details")
for i in b:
    c=i.find("h2",class_="title").text.strip()
    name.append(c)
    m=i.find("div",class_="mileage").text.strip()
    mileage.append(m)
    d=i.find("div",class_="dealer-name").text.strip()
    dealer_name.append(d)
    p=i.find("span",class_="primary-price").text.strip()
    price.append(p)
    r=i.find("span",class_="sds-rating__link sds-button-link").text.strip()
    review.append(r[1:-9])
    try:
        rat=i.find("span",class_="sds-rating__count").text.strip()
        rating.append(rat)
    except:
        rating.append(np.nan)
df=pd.DataFrame({"Carname":name,"mileage":mileage,"dealername":dealer_name,"price":price,"review":review,"rating":rating})
df
```

```python
from bs4 import BeautifulSoup
import pandas as pd
website="30 Most Popular Movies Right Now_ What to Watch In Theaters and Streaming __ Rotten Tomatoes – Movie and TV News.html"
file=open(website,encoding="utf-8")
response=file.read()
soup=BeautifulSoup(response,"html.parser")
movie=[]
rating=[]
director=[]
a=soup.find_all("div",class_="col-sm-18 col-full-xs countdown-item-content")
for i in a:
    movie.append(i.find("h2").find("a").text.strip())
    rating.append(i.find("span",class_="tMeterScore").text.strip())
    director.append(i.find("div",class_="info director").find("a").text.strip())
df=pd.DataFrame({"movie":movie,"rating":rating,"director":director})
df
```

```python
from bs4 import BeautifulSoup
import pandas as pd
website="SENSEX Top Gainers, BSE Daily Top Gainers - Stock Market (1).html"
file=open(website,encoding="utf-8")
response=file.read()
soup=BeautifulSoup(response,"html.parser")
a=soup.find("table")
b=soup.find_all("tr")
company=[]
price=[]
c_date=[]
change=[]
change_pr=[]
day=[]
for i in b[1:]:
    company.append(i.find("td",class_="v-align").text.strip())
    price.append(i.find("td",class_="txt-right v-align").text.strip())
    c_date.append(i.find("td",class_="txt-right no-wrap").find("small").text.strip())
    days=i.find("div",class_="btm clearfix").text.strip()
    d1=days.split("\n")
    d2=d1[0]+" "+d1[1]
    day.append(d2)
    c=i.find_all("td",class_="txt-ctr")
    for i in c[::2]:
        change.append(i.text.strip())
    for i in c[1::2]:
        change_pr.append(i.text.strip())
df=pd.DataFrame({"Company":company,"Current Price":price,"Prev Close Date":c_date,"Change":change,"Change %":change_pr,
                 "	Day's Low/High":day})
df
```
